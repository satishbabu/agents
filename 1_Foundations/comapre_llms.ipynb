# Compare Open AI, Claude and Gemini
This will help learn to connect to various LLMs and it will test that setup for all LLMs are working well

```python
import os
import json
from dotenv import load_dotenv
from openai import OpenAI
from anthropic import Anthropic
from IPython.display import Image, display, Markdown

load_dotenv(override=True)

open_api_key = os.getenv('OPENAI_API_KEY')
anthropic_api_key = os.getenv('ANTHROPIC_API_KEY')
google_api_key = os.getenv('GOOGLE_API_KEY')

if open_api_key:
    print("OpenAI API key is set")
else:
    print("OpenAI API key is not set")

if anthropic_api_key:
    print("Anthropic API key is set")
else:
    print("Anthropic API key is not set")

if google_api_key:
    print("Google API key is set")
else:
    print("Google API key is not set")

# Test OpenAI
```

```python
question = "Assume you are a travel agent.  A family of 4 wants to travel to Spain to visit Barcelona and Madrid. "
question += "The family is interested in history, culture and food.  The family consists of 2 adults and 2 teenage children.  Prepare a 10 day itinerary for them. "
question += "Ensure the itinerary is not too crowded and the family is able to enjoy the trip.  The father is budget conscious and wants to save money. "
question += "The family is also interested in the art and the architecture of the places they are visiting.  "
question += "The family is also interested in the music and the dance of the places they are visiting.  "
question += "The family prefers airbnb in the city center to experience the city and save time travelling.  "
question += "Esnure to include the total cost that includes the accomodation, food, sightseeing, trasportation, etc. "

models = []
answers = []
messages = [{'role': 'user', 'content': question}]
```

```python

model = "gpt-5-nano"

openai = OpenAI()
response = openai.chat.completions.create(model=model, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
models.append(model)
answers.append(answer)

```

```python
print(messages)
print(len(answers))
print(models)
```

```python
model = "claude-sonnet-4-5"

claude = Anthropic()
response = claude.messages.create(model=model, messages=messages, max_tokens=1000)
answer = response.content[0].text

display(Markdown(answer))
models.append(model)
answers.append(answer)
```

```python
gemini = OpenAI(api_key=google_api_key, base_url="https://generativelanguage.googleapis.com/v1beta/openai/")
model = "gemini-2.5-flash"

response = gemini.chat.completions.create(model=model, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
models.append(model)
answers.append(answer)
```

```python
judge = f"""You are judging a competition between {len(models)} competitors.
Each model has been given this question:

{question}

Your job is to evaluate each response for clarity and strength of argument, and rank them in order of best to worst.
Respond with JSON, and only JSON, with the following format:
{{"results": ["best competitor number", "second best competitor number", "third best competitor number", ...]}}

Here are the responses from each competitor:

# Response from competitor 1 \n\n
{answers[0]}



# Response from competitor 2 \n\n
{answers[1]}



# Response from competitor 3 \n\n
{answers[2]}



Now respond with the JSON with the ranked order of the competitors, nothing else. Do not include markdown formatting or code blocks."""

```

```python
judge_messages = [{"role": "user", "content": judge}]
print(judge_messages)
```

```python
# compare the answers using openAI

from unittest import result


openai = OpenAI()
response = openai.chat.completions.create(
    model = "gpt-5-mini",
    messages=judge_messages
)
llm_ranking = response.choices[0].message.content
print(llm_ranking)
```

```python
# OK let's turn this into results!

results_dict = json.loads(llm_ranking)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    model = models[int(result)-1]
    print(f"Rank {index+1}: {model}")
```
