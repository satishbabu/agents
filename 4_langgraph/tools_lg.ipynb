### First, let's go set up LangSmith!

1. Signup at https://langsmith.com.  Get the 4 env variables and add it to .env.  One of the env variable is key and ensure it is not exposed.

2. Signup for https://serper.dev.  It is a wrapper arround google search.  Get the API key and add it to the .env as SERPER_API_KEY

```python
# Do a google search using Serper API and check the logs on langsmith.com
from langchain_community.utilities import GoogleSerperAPIWrapper

serper = GoogleSerperAPIWrapper()
serper.run("What is the capital of USA")
```

```python
# create a tool for serper function using langchain wrapper class

from langchain.agents import Tool

tool_search =Tool(
        name="search",
        func=serper.run,
        description="Useful for when you need more information from an online search"
    )
```
